# GOES 12: reproduce Figure 2's light curve

This notebook reproduces one within-exposure light curve from public IRSA science and mask cutouts. Install `glint-geo` in this Python environment first. Historical orbital elements require your own Space-Track account; set `SPACETRACK_USER` and `SPACETRACK_PASSWORD` in your shell before launching Jupyter. The notebook stores no elements or credentials.

The public exposure ID is 115845983 (2020-03-03 Palomar night; 2020-03-04 UTC). The extracted magnitudes use the ZTF header field-star zero point and have no satellite color correction. The plot defaults to median-relative magnitude with an SNR 3 display cutoff.

```python
import os
import subprocess
import sys
from pathlib import Path

if not (os.environ.get('SPACETRACK_USER') and os.environ.get('SPACETRACK_PASSWORD')):
    raise RuntimeError('Historical GOES 12 data require Space-Track environment credentials')

def run(*args):
    subprocess.run([sys.executable, '-m', 'glint_geo.cli', *args], check=True)

run('fetch', '--norad-id', '26871', '--start', '2020-03-04T11:02:00Z',
    '--end', '2020-03-04T11:03:00Z', '--exposure-id', '115845983', '--limit', '1')
run('extract')
```

```python
curve = Path('out/26871_115845983_c07q2.csv')
run('plot', '--curve', str(curve), '--output', 'out/goes12_figure2.png')
from IPython.display import Image, display
display(Image(filename='out/goes12_figure2.png'))
```
