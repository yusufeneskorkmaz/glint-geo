# GOES 12: one public ZTF streak light curve

This notebook reproduces one within-exposure light curve from public IRSA science and mask cutouts. Install `glint-geo` in this Python environment first. Historical orbital elements require your own Space-Track account; set `SPACETRACK_USER` and `SPACETRACK_PASSWORD` in your shell before launching Jupyter. The notebook stores no elements or credentials.

The public exposure ID is 89520732 (2019-06-15 UTC). The extracted magnitudes use the ZTF header field-star zero point and have no satellite color correction.

```python
import os
import subprocess
import sys
from pathlib import Path

if not (os.environ.get('SPACETRACK_USER') and os.environ.get('SPACETRACK_PASSWORD')):
    raise RuntimeError('Historical GOES 12 data require Space-Track environment credentials')

def run(*args):
    subprocess.run([sys.executable, '-m', 'glint_geo.cli', *args], check=True)

run('fetch', '--norad-id', '26871', '--start', '2019-06-15T04:58:00Z',
    '--end', '2019-06-15T05:00:00Z', '--exposure-id', '89520732', '--limit', '1')
run('extract')
```

```python
curve = Path('out/26871_89520732_c05q2.csv')
run('plot', '--curve', str(curve), '--output', 'out/goes12.png')
from IPython.display import Image, display
display(Image(filename='out/goes12.png'))
```
